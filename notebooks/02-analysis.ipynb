```python
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

# Загружаем данные (путь без ../, так как ядро запущено из корня)
df = pd.read_csv("fish_participant.csv")

print("Размер датасета:", df.shape)
print("\nСредний вес по видам:")
print(df.groupby("Species")["Weight"].mean().sort_values(ascending=False).round(1))
```

```python
# Показываем график остатков
img = Image.open("reports/figures/residuals_plot.png")
plt.figure(figsize=(10, 6))
plt.imshow(img)
plt.axis("off")
plt.show()
```

Выводы по анализу
1. Самый тяжелый вид — Pike (щука), самый легкий — Smelt (корюшка).
2. График остатков показывает, что модель лучше работает на мелких рыбах (MAE = 23 г) и хуже на крупных (MAE = 60 г).
3. Лучший результат показал Random Forest (R² = 0.97), так как он умеет находить нелинейные связи между размерами и весом.
